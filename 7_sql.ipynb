```sql
CREATE DATABASE my_database;
USE my_database; 


```SQL
CREATE TABLE customers (
    customer_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(50),
    age INT,
    city VARCHAR(50)
);


```sql
INSERT INTO customers (name, age, city) 
VALUES ('Alice', 30, 'New York'),
       ('Bob', 25, 'Chicago'),
       ('Carol', 28, 'Los Angeles');


```sql
SELECT * FROM customers;


___

## Filtering data

```sql
SELECT * FROM customers WHERE city = 'New York';


```sql
SELECT * FROM customers WHERE age > 25 AND city = 'Chicago';
SELECT * FROM customers WHERE age BETWEEN 20 AND 30;
SELECT * FROM customers WHERE name LIKE 'A%'; -- Names starting with 'A'


## update & delete

```sql
UPDATE customers SET age = 31 WHERE name = 'Alice';
DELETE FROM customers WHERE name = 'Bob';


## 6. Sorting & Limiting Results

```sql
SELECT * FROM customers ORDER BY age DESC; -- Sort by age (highest first)
