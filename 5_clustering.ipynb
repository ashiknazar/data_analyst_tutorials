# Step 2: Start a Dask Scheduler (on the main device)

`dask-scheduler`

- this will output an ip address and port ,something like:
  - Scheduler at: tcp://192.168.1.100:8786

# Step 3: Start Dask Workers (on both devices)

- Now, connect workers to the scheduler.
- **On Device 1 (Main Device)**

`dask-worker tcp://192.168.1.100:8786`


- On Device 2 (Second Device)
- Run this on Device 2, replacing the IP with the scheduler’s IP:

`dask-worker tcp://192.168.1.100:8786` 


- Once connected, it will show something like:
- Registered to scheduler at: tcp://192.168.1.100:8786


# Step 4: Connect from Python Code
- You can now use the cluster in Python. In Jupyter Notebook or Python script, run:

```python
from dask.distributed import Client
client = Client("tcp://192.168.1.100:8786")  # Use the scheduler's IP
print(client)

This will display details about the connected cluster.

# Step 5: Run a Test Task
Try running a test computation:

```python
import dask.array as da

arr = da.random.random((10000, 10000), chunks=(1000, 1000))
result = arr.mean().compute()
print(result)
